One Input One Output (Simple Linear Regression)

This only work for one input one output applications

```python
import pandas as pd
import numpy as np 
```

```python
class customLR:
    
    def __init__(self):
        self.m = None
        self.b = None
        
    def fit(self,X_train,y_train):
        
        num = 0
        den = 0
        
       
            
        num =  ((X_train - X_train.mean())*(y_train - y_train.mean())).sum()
        den =  ((X_train - X_train.mean())*(X_train - X_train.mean())).sum()
        
        self.m = num/den
        self.b = y_train.mean() - (self.m * X_train.mean())
        print(self.m)
        print(self.b)       
    
    def predict(self,X_test):
        
        print(X_test)
        
        return self.m * X_test + self.b
```

```python
model=customLR()
```

```python
df=pd.read_csv(r'simple_linear_regression_dataset.csv')
```
